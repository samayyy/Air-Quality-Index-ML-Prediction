# 01 - Data Download & Initial Inspection

**Dataset**: Air Quality Data in India (2015-2020) by Rohan Rao (Kaggle)

**Files**:
- `city_day.csv` - Daily city-level aggregated data (primary)
- `city_hour.csv` - Hourly city-level data
- `station_day.csv` - Daily station-level data
- `station_hour.csv` - Hourly station-level data
- `stations.csv` - Station metadata

```python
import pandas as pd
import numpy as np
import os
import sys

# Add project root to path
sys.path.insert(0, os.path.abspath('..'))
from config import DATA_RAW, POLLUTANT_COLS

print(f"Data directory: {DATA_RAW}")
print(f"Files available: {os.listdir(DATA_RAW)}")
```

## 1. City Day Data (Primary Dataset)

```python
city_day = pd.read_csv(DATA_RAW / 'city_day.csv', parse_dates=['Date'])
print(f"Shape: {city_day.shape}")
print(f"\nDate range: {city_day['Date'].min()} to {city_day['Date'].max()}")
print(f"Cities: {city_day['City'].nunique()} - {sorted(city_day['City'].unique())}")
city_day.head(10)
```

```python
city_day.info()
```

```python
city_day.describe().round(2)
```

```python
# Missing values summary
missing = city_day.isnull().sum()
missing_pct = (missing / len(city_day) * 100).round(2)
missing_df = pd.DataFrame({'Count': missing, 'Percentage': missing_pct})
missing_df = missing_df[missing_df['Count'] > 0].sort_values('Percentage', ascending=False)
print("Missing Values Summary:")
print(missing_df.to_string())
```

```python
# AQI Bucket distribution
print("\nAQI Bucket Distribution:")
print(city_day['AQI_Bucket'].value_counts())
print(f"\nAQI missing: {city_day['AQI'].isna().sum()} ({city_day['AQI'].isna().mean()*100:.1f}%)")
```

## 2. Station Day Data

```python
station_day = pd.read_csv(DATA_RAW / 'station_day.csv', parse_dates=['Date'])
print(f"Shape: {station_day.shape}")
print(f"Stations: {station_day['StationId'].nunique()}")
print(f"Date range: {station_day['Date'].min()} to {station_day['Date'].max()}")
station_day.head(5)
```
